# Modeling Pipeline
This notebook covers:
- Preprocessing
- Baseline model
- Model experiments
- Hyperparameter tuning
- Evaluation
- Submission file generation

```python
# Add xgboost dependency for gradient boosting
!pip install xgboost

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, roc_auc_score, roc_curve
import matplotlib.pyplot as plt

TRAIN_URL = "https://raw.githubusercontent.com/Arthur-IO/Data-Science-Final-Proj/master/data/train.csv"
TEST_URL = "https://raw.githubusercontent.com/Arthur-IO/Data-Science-Final-Proj/master/data/test.csv"
SUB_URL  = "https://raw.githubusercontent.com/Arthur-IO/Data-Science-Final-Proj/master/data/sample_submission.csv"

train = pd.read_csv(TRAIN_URL)
test = pd.read_csv(TEST_URL)
sample_sub = pd.read_csv(SUB_URL)

target = "Target"
X = train.drop(target, axis=1)
y = train[target]

le = LabelEncoder()
y_encoded = le.fit_transform(y)

# find which columns are categories and which are numbers
categorical_cols = X.select_dtypes(include="object").columns
numerical_cols = X.select_dtypes(exclude="object").columns

numerical_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# combine boh pipelines so the model can use all features
preprocess = ColumnTransformer([
    ("num", numerical_pipe, numerical_cols),
    ("cat", categorical_pipe, categorical_cols)
])

X_train, X_valid, y_train, y_valid = train_test_split(
    X, y_encoded, test_size=0.2, stratify=y_encoded, random_state=42
)
```

# Baseline Model: Logistic Regression


```python
logreg = LogisticRegression(max_iter=1000, class_weight="balanced")
logreg.fit(preprocess.fit_transform(X_train), y_train)

y_pred = logreg.predict(preprocess.transform(X_valid))
y_proba = logreg.predict_proba(preprocess.transform(X_valid))

print(classification_report(y_valid, y_pred))

# calculate ROC-AUC score to measure how well the model separates classes
print("ROC-AUC (ovr):", roc_auc_score(y_valid, y_proba, multi_class="ovr"))
```

# Experiment with random forest and xgboost models


```python
rf = RandomForestClassifier(n_estimators=300, class_weight="balanced", random_state=42)
rf.fit(preprocess.fit_transform(X_train), y_train)

from xgboost import XGBClassifier
xgb = XGBClassifier(n_estimators=500, learning_rate=0.05, max_depth=6, random_state=42)
xgb.fit(preprocess.fit_transform(X_train), y_train)

models = {"LogReg": logreg, "RandomForest": rf, "XGB": xgb}
for name, model in models.items():
    y_proba = model.predict_proba(preprocess.transform(X_valid))
    print(name, "ROC-AUC (ovr):", roc_auc_score(y_valid, y_proba, multi_class="ovr"))
```

# Quick Checkpoint Summary

## Preprocessing
- Numerical features: median imputation + standard scaling
- Categorical features: most frequent imputation + one-hot encoding
- 80-20 train-val split with stratification

## Model Results
- **Logistic Regression**: 80% accuracy, 0.922 class separation score
- **Random Forest**: 0.931 class separation score
- **XGBoost**: 0.936 class separation score (best so far)

Tree models > linear models for this dataset. XGBoost currently winning. Class 1 (Enrolled) is the hardest to predict correctly.

# RandomizedSearchCV for Random Forest
Tune hyperparameters systematically.

```python
from sklearn.model_selection import RandomizedSearchCV

param_dist = {
    "n_estimators": [200, 400, 600],
    "max_depth": [None, 8, 12],
    "min_samples_split": [2, 5, 10]
}

search = RandomizedSearchCV(rf, param_dist, n_iter=5, cv=3, scoring="accuracy", n_jobs=-1)
search.fit(preprocess.fit_transform(X_train), y_train)
print(search.best_params_)
```

# Compare metrics
Build a summary table of model performance.

```python
import pandas as pd

results = []
for name, model in models.items():
   y_pred = model.predict(preprocess.transform(X_valid))
   y_proba = model.predict_proba(preprocess.transform(X_valid))
   results.append({
       "Model": name,
       "Accuracy": (y_pred == y_valid).mean(),
       "ROC-AUC": roc_auc_score(y_valid, y_proba, multi_class="ovr")
   })

pd.DataFrame(results)
```

```python
# ROC Curves for each class (One-vs-Rest approach)
# For multiclass problems, we plot separate ROC curves for each class

from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc

# Binarize the labels for one-vs-rest ROC calculation
classes = [0, 1, 2]
class_names = ["Dropout", "Enrolled", "Graduate"]
y_valid_bin = label_binarize(y_valid, classes=classes)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for idx, (class_id, class_name) in enumerate(zip(classes, class_names)):
    ax = axes[idx]
    
    for name, model in models.items():
        y_proba = model.predict_proba(preprocess.transform(X_valid))
        fpr, tpr, _ = roc_curve(y_valid_bin[:, class_id], y_proba[:, class_id])
        roc_auc = auc(fpr, tpr)
        ax.plot(fpr, tpr, label=f"{name} (AUC = {roc_auc:.3f})")
    
    ax.plot([0, 1], [0, 1], "k--", label="Random Guess")
    ax.set_xlabel("False Positive Rate")
    ax.set_ylabel("True Positive Rate")
    ax.set_title(f"ROC Curve - {class_name}")
    ax.legend(loc="lower right")

plt.tight_layout()
plt.show()
```

# Predictions on Test Set
Using our best model (XGBoost) to predict academic outcomes for unseen students.

```python
# Make predictions on test set using XGBoost (best model)
test_predictions = xgb.predict(preprocess.transform(test))

# Convert numeric predictions back to labels
prediction_labels = le.inverse_transform(test_predictions)

# Show some sample predictions
sample_results = pd.DataFrame({
    "Student ID": test["id"].head(10),
    "Predicted Outcome": prediction_labels[:10]
})
print("Sample predictions for first 10 students:")
print(sample_results)

# Show distribution of predictions
print("\nPrediction distribution across all test students:")
print(pd.Series(prediction_labels).value_counts())
```

# Conclusion

After testing three different models, XGBoost came out on top with 83% accuracy and a 0.936 ROC-AUC score. Random Forest was close behind at 0.931, and Logistic Regression trailed at 0.922. Overall, the tree-based models worked much better than the linear approach for this problem.

One interesting finding was that "Enrolled" students were the hardest to predict correctly. Looking at the ROC curves, the AUC for that class was only around 0.89 compared to 0.95+ for Dropout and Graduate. This probably happens because students who are still enrolled share traits with both groups - some will eventually drop out and others will graduate.

The preprocessing pipeline handled the data well using median imputation and scaling for numbers, and one-hot encoding for categories. We used an 80-20 train-validation split with stratification to make sure each class was represented fairly. For tuning, we ran RandomizedSearchCV on Random Forest with 3-fold cross-validation which helped us find better hyperparameters.

All three models achieved ROC-AUC scores above 0.92, which shows the dataset features are pretty informative for predicting student outcomes.
